### Opgave 3
Vi vil gerne prøve at lave en fysisk model for skiven. Til starte med vil vi antage at accretion rate er så lav at temperaturen i skiven er bestemt af stjernens temperatur. Så kan man finde temperaturen af støvet i midtplanet ved hjælp af formlen:
$$ T_{mid}(r) = \left( \frac{\varphi L_*}{8\pi r^2 \sigma_{SB}}\right)^{1/4}\, $$
hvor $\varphi$ er forholdet imellem højden og radius på skiven. Vi vil antage at $\varphi=0.02$.

Lav et plot for temperatur profilen af stjernen. Hvad er det typiske temperatur område for din skive?

Hvor gennemsigtig skiven er ved en given radius og bølgelængde afhænger af hvor ugennemsigtig støvet er. Hvis støvet det ikke er fuldstændigt ugennemsigtigt siger man at det er _optisk tyndt_. Opaciteten angiver hvor ugennemsigtigt enten gas eller støv er for lys ved en given bølgelængde og måles i [ cm$^2$ $g^{-1}$ ]. Observationerne er lavet ved en bølgelængde på $\lambda=1.25$ mm, hvilket svarer til en frekvens på $\nu=240$ GHz. For støv i DSHARP skiverne ved 1.25 mm gælder der at opciteten er ca $\kappa_\textrm{dust} = 2$ cm$^2$ $g^{-1}$. Hvis vi kender opaciteten kan vi måle den _optiske dybde_, $\tau$, som
$$ \tau_\nu = \kappa_\nu \Sigma_d(r) $$
hvor $\Sigma_d(r)$ er overflade tætheden af støvet. Den optiske dybde er et udtryk for hvor meget stråling der er gået tabt undervejs. Man siger at noget er optisk tykt hvis $\tau=1$. Hvis vi kender temperaturen af midtplanet kan vi antage at det stråler som et sort legeme. Vi har også målt intensiteten af lyset i observationerne, og derfor kan vi sammenligne de to til at finde ud af hvad den optiske dybde er. Der gælder:
$$ I_\nu = B_\nu(T) ( 1 - \exp(-\tau_\nu) )\,, $$
hvor $B_\nu(T)$ er sortlegeme intensiteten. Hvis vi indsætter $\nu = 240 GHz$ finder vi at
$$ B_\nu(T) = \frac{2 h\nu^3}{c^2} \frac{1}{\exp\left({h\nu \over k_B T}\right) - 1}
= 0.48\,\textrm{J as}^{-2}\frac{1}{\exp\left({11.5 \textrm{K} \over T}\right) - 1} $$

Brug udtrykkene til at lave en graf for den optiske dybde og for støv overflade tætheden. Er skiven optisk tyk nogen steder; hvor er skiven tynd henne, og hvorfor?

Nu hvor I har fundet en støv overflade tæthed kan I beregne den total støv masse. Antag at støv til gas forholdet er 1:100 og beregn også en gas masse. Er det rimeligt i forhold til stjernens masse, og er der nok støv til at lave planeter? (hvad er gode enheder for masserne?)

```python
Lsun = 3.828e26 # Watt
Msun = 1.98847e33 # g
Mjup = 1.89813e30 # g
au_in_m = 149597870700
sigma_SB = 5.67e-8 # W / m^2 K^-4

Mdot  = 10**(-6.4) # accretion rate
Mstar = 10**(-0.11) # stellar mass
Tstar=10**3.63 # stellar surface temperature
Lstar=10**(0.78)*Lsun # stellar luminosity
varphi = 0.02
Tmid = (varphi * Lstar / (8*np.pi*sigma_SB*(r_au*au_in_m)**2))**0.25

xdata = r_au
ydata = Tmid
f = plt.figure(figsize=(6, 5), dpi=150, facecolor='w', edgecolor='k')
plt.plot(xdata,ydata,marker='o',markersize=2.0,linewidth=0.5)
plt.yscale('log')
plt.xlabel('Radius [AU]')
plt.ylabel('Midplane temperature [ K ]')
plt.show()

Bnu = 0.48 / (np.exp(11.5 / Tmid) - 1.)
tau = -np.log(1. - Inu / Bnu)
kappa = 2. # cm^2 / g
sigma_d = tau / kappa

dust_to_gas = 100.
sigma_gas = dust_to_gas * sigma_d

from scipy.integrate import simps
r_cm = r_au * au_in_m * 1e2
Mdisk_d = simps(sigma_d * 2. * np.pi * r_cm, x=r_cm)
Mdisk_gas = Mdisk_d

xdata = r_au
ydata = tau
f = plt.figure(figsize=(6, 5), dpi=150, facecolor='w', edgecolor='k')
plt.plot(xdata,ydata,marker='o',markersize=2.0,linewidth=0.5)
plt.yscale('log')
plt.xlabel('Radius [AU]')
plt.ylabel('Optical depth')
plt.show()

xdata = r_au
ydata = sigma_d
f = plt.figure(figsize=(6, 5), dpi=150, facecolor='w', edgecolor='k')
plt.plot(xdata,ydata,marker='o',markersize=2.0,linewidth=0.5)
plt.yscale('log')
plt.xlabel('Radius [AU]')
plt.ylabel('Surface density [g cm$^{-2}$]')
plt.show()

print("Dust disk mass [Jupiter mass]:", Mdisk_d / Mjup)
print("Gas disk mass [Solar mass]:", Mdisk_gas / Msun)
print("Gas disk mass [Stellar mass]:", Mdisk_gas / (Mstar * Msun))
```

### Opgave 4
Nu da vi kender temperatur, støv og gas fordelingen i skiven kan vi prøve at beregne $\alpha$-parameteren der måler hvor kraftig turbulensen er. Vi vil gøre det ved at lave en meget grov. antagelse, nemlig at _accretion_rate_ på stjernen er konstant ind igennem skiven. Så er det muligt -- ved at bruge resultater fra bogen og fra slides -- at beregne en effektiv alpha parameter som funktion a radius.

Hvis I har valgt en skive med gaps / feautures vil I nok finde at $\alpha$ stiger i gaps. Hvorfor? Er det en rimelig størrelse I finder for $\alpha$?

I kender accretion rate og skive massen, og hvis I antager at skiven ikke har nogen gas-kerne der kan tilføre nyt materiale, hvor lang er da levetiden på skiven. Passer det med stjernens alder?

```python
G = 6.67430e-8 # cgs units
Omega_K = np.sqrt(G * Mstar * Msun / r_cm**3)
cs = 1.9e5 * (Tmid / 10.)**0.5 # cm/s
yr = 365.25 * 24 * 3600. # year in seconds
Mdot_cgs = Mdot * Msun / yr # g / s accretion rate
alpha = Omega_K / cs**2 * Mdot_cgs / (3. * np.pi * sigma_gas)

xdata = r_au
ydata = alpha
f = plt.figure(figsize=(6, 5), dpi=150, facecolor='w', edgecolor='k')
plt.plot(xdata,ydata,marker='o',markersize=2.0,linewidth=0.5)
plt.yscale('log')
plt.xlabel('Radius [AU]')
plt.ylabel('Alpha parameter')
plt.show()

print("projekteret leve-tid [yr]", int(Mdisk_gas / Mdot_cgs / yr))
# AS209 har en ekstrem kort levetid. Der må enten være noget galt med
# støv-gas forholdet, eller også gemmer der sig en masse materiale, da
# skiven er optisk tyk, eller en 3. mulighed er at systemet alligvel er
# omgivet af en kerne.
```
