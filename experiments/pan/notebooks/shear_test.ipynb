```python
io=8
a=8; w=1; x0=2+2*0.1*io
run='shear_hllc'
figure(2); clf()

subplot(2,2,1)
sn=dispatch.snapshot(io,run,'../data/')
dgr.amr_plane(sn,z=0.5,iv=0,log=0,mesh=1)
xlim(0.,8); ylim(0.,3)

subplot(2,2,2)
dgr.amr_plane(sn,z=0.5,iv='uy',log=0,mesh=1,cmap='coolwarm')
xlim(x0-w,x0+w); ylim(0.,1)

#figure(2); clf()
subplot(2,2,3)
for i in range(8):
    p=sn.patches[i]
    semilogy(p.xi,p.var('d')[:,8,8],'-+')
title1='hllc-slope=3.3-a={}-u0=2'.format(a)
title(title1)
subplot(2,2,4)
ylim(-a-1,a+3)
xlim(x0-w,x0+w)
for i in range(8):
    p=sn.patches[i]
    plot(p.xi,p.var('ux')[:,8,8],'-+')
    plot(p.xi,4.*p.var('uy')[:,8,8],'-+')
tight_layout()
title('t={:.3f}'.format(sn.time))
savefig(title1+'.png')
```
