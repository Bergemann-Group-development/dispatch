```python
def Show(run='mhd',io=10,z=0.4):
    sn=dis.snapshot(io,run)
    figure(figsize=(15,4))
    subplot(1,3,1); dgr.unigrid_plane(sn,'lnd',z=z)
    subplot(1,3,2); dgr.unigrid_plane(sn,'ux',z=z)
    subplot(1,3,3); dgr.unigrid_plane(sn,'bz',z=z)
    tight_layout()
Show('hydro',5)
Show('hydro')
```
