# Showing a slice perpendicular to the y-axis

### Using dispatch.graphics

```python
sn=dispatch.snapshot(1,run=run,data=data)
f=dse.unigrid_plane(sn,iv='qr',y=0.5)
figure(2,figsize=(12,4))
clf()
subplot(1,2,1)
dgr.imshow(f**0.15)
subplot(1,2,2)
# Show left/right symmetry
plot(f[3,:],'-+')
plot(f[-4,:],'-+');
```

### Using dispatch.yt and YT

```python
c=[0.5,0.5,0.5]; w=((1,'cm'),(1,'cm'))
v='qr'
ds=dyt.open_amr(iout=2,run=run,data=data,copy=True)
slc=yt.SlicePlot(ds=ds,axis="y",fields=[v],center=c,width=w)
slc.set_cmap(v,"inferno")
slc.show()
```
