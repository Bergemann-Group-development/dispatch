## Image slices

Log density in the z=0 plane:

```python
figure(figsize(20,4))
for io in range(1,3):
    sn=dis.snapshot(io,run,data); subplot(1,4,io)
    dgr.amr_plane(sn,'logd',width=250)
```

Shear (y-) velocity in xy-plane

```python
figure(figsize(20,4))
for io in range(1,3):
    sn=dis.snapshot(io,run,data); subplot(1,4,io)
    dgr.amr_plane(sn,'uy',width=4500,zero=1)
```

Log density in yz-plane

```python
figure(figsize(20,4))
for io in range(1,3):
    sn=dis.snapshot(io,run,data); subplot(1,4,io)
    dgr.amr_plane(sn,'logd',width=4500,x=0,vmax=-11)
```

Vertical velocity in full yz-plane -- __there is boundary condition issue at the lower boundary__

```python
figure(figsize(20,4))
for io in range(1,3):
    sn=dis.snapshot(io,run,data); subplot(1,4,io)
    dgr.amr_plane(sn,'uz',width=4500,x=0,zero=True)
```

Vertical velocity in 250 width yz-plane 

```python
figure(figsize(20,4))
for io in range(1,3):
    sn=dis.snapshot(io,run,data); subplot(1,4,io)
    dgr.amr_plane(sn,'uz',width=250,x=0)
```

Vertical velocity in 120 width yz-plane 

```python
figure(figsize(20,4))
for io in range(1,3):
    sn=dis.snapshot(io,run,data); subplot(1,4,io)
    v=0.01
    dgr.amr_plane(sn,'uz',width=120,x=0,vmin=-v,vmax=v)
```

... with patch-mesh

```python
figure(figsize(20,4))
for io in range(1,3):
    sn=dis.snapshot(io,run,data); subplot(1,4,io)
    v=0.01
    dgr.amr_plane(sn,'uz',width=120,x=0,vmin=-v,vmax=v,mesh=1)
```
