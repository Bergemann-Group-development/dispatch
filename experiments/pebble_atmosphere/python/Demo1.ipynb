## Image slices

Log density in the z=0 plane:

```python
figure(figsize(20,4))
for io in range(1,5):
    sn=dis.snapshot(io,run,data); subplot(1,4,io)
    dgr.amr_plane(sn,'logd',width=10,vmax=-8,vmin=-10)
```

... with mesh visible:

```python
figure(figsize(20,4))
for io in range(1,5):
    sn=dis.snapshot(io,run,data); subplot(1,4,io)
    dgr.amr_plane(sn,'logd',width=20,vmax=-8,vmin=-10,mesh=1)
```

```python
v=.3
figure(figsize(20,4))
for io in range(1,5):
    sn=dis.snapshot(io,run,data); subplot(1,4,io)
    dgr.amr_plane(sn,'ux',width=30,vmin=-v,vmax=v)
```

Radial velocity in the z=0 plane:

```python
v=.05
figure(figsize(20,4))
for io in range(1,5):
    sn=dis.snapshot(io,run,data); subplot(1,4,io)
    dgr.amr_plane(sn,'ux',width=500,vmin=-v,vmax=v)
```

Azimuthal shear in the z=0 plane:

```python
v=.4
figure(figsize(20,4))
for io in range(1,5):
    sn=dis.snapshot(io,run,data); subplot(1,4,io)
    dgr.amr_plane(sn,'uy',width=500,vmin=-v,vmax=v)
```

```python
v=10
figure(figsize(20,4))
for io in range(1,5):
    sn=dis.snapshot(io,run,data); subplot(1,4,io)
    dgr.amr_plane(sn,'uy',y=-1700*(6+2.5),width=11000,vmin=-v,vmax=v,mesh=1,ident=1)
```

```python
v=0.5
figure(figsize(20,4))
for io in range(1,5):
    sn=dis.snapshot(io,run,data); subplot(1,4,io)
    dgr.amr_plane(sn,'ux',y=-1700*(6+2.5),width=11000,vmin=-v,vmax=v)
```

Vertical velocity in the y=0 plane:

```python
v=.3
figure(figsize(20,4))
for io in range(1,5):
    sn=dis.snapshot(io,run,data); subplot(1,4,io)
    dgr.amr_plane(sn,'uz',x=0,width=30,vmin=-v,vmax=v)
```

Velocity amplitude in the xz-plane at y=0

```python
v=.3
figure(figsize(20,4))
for io in range(1,5):
    sn=dis.snapshot(io,run,data); subplot(1,4,io)
    dgr.amr_plane(sn,'(ux**2+uz**2)**0.5',y=0,width=30,vmin=-v,vmax=v)
```
